```python
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Load Zachary's Karate Club Graph
G = nx.karate_club_graph()
m = G.number_of_edges()
nodes_list = list(G.nodes())
A = nx.to_numpy_array(G, nodelist=nodes_list)
k = np.sum(A, axis=1)

# Full graph Modularity Matrix B
B_full = A - np.outer(k, k) / (2 * m)

def split_community(sub_indices):
    """
    Attempts to split a subgraph defined by `sub_indices` using the leading eigenvector
    of the generalized modularity matrix B^(g).
    Returns two lists of node indices if delta_Q > 0, else None.
    """
    if len(sub_indices) <= 1:
        return None
    
    # Submatrix of B for the current community
    B_sub = B_full[np.ix_(sub_indices, sub_indices)]
    
    # Generalized modularity matrix B^(g): subtract row sums from diagonal
    B_g = B_sub - np.diag(np.sum(B_sub, axis=1))
    
    # Compute eigenvalues and eigenvectors (B_g is symmetric, so use eigh)
    eigenvalues, eigenvectors = np.linalg.eigh(B_g)
    
    # Leading eigenvalue and corresponding eigenvector
    max_idx = np.argmax(eigenvalues)
    max_val = eigenvalues[max_idx]
    leading_vec = eigenvectors[:, max_idx]
    
    # Stop if no positive eigenvalue exists
    if max_val <= 1e-10:
        return None
    
    # Assign +1 or -1 based on sign of leading eigenvector
    s = np.where(leading_vec > 0, 1, -1)
    
    # Calculate modularity contribution delta_Q
    delta_Q = (s.T @ B_g @ s) / (4 * m)
    
    if delta_Q <= 1e-10:
        return None
    
    group_pos = [sub_indices[i] for i in range(len(sub_indices)) if s[i] == 1]
    group_neg = [sub_indices[i] for i in range(len(sub_indices)) if s[i] == -1]
    
    if len(group_pos) == 0 or len(group_neg) == 0:
        return None
        
    return group_pos, group_neg

# Run Recursive Partitioning and record history at each iteration
communities = [list(range(len(nodes_list)))]
history = [[list(c) for c in communities]] # Iteration 0: whole graph

while True:
    next_communities = []
    split_occurred = False
    
    for comm in communities:
        result = split_community(comm)
        if result is not None:
            next_communities.extend(result)
            split_occurred = True
        else:
            next_communities.append(comm)
            
    if not split_occurred:
        break
        
    communities = next_communities
    history.append([list(c) for c in communities])

print(f"Total iterations (splits): {len(history) - 1}")
print(f"Final number of communities detected: {len(communities)}")
for idx, c in enumerate(communities):
    print(f"Community {idx + 1}: {[nodes_list[i] for i in c]}")
    # Fix layout so node positions stay consistent across iterations
pos = nx.spring_layout(G, seed=42)
cmap = plt.get_cmap('tab10', 10)
fig, axes = plt.subplots(1, len(history), figsize=(6 * len(history), 5))
if len(history) == 1:
    axes = [axes]

for step, comms in enumerate(history):
    ax = axes[step]
    node_colors = [0] * len(nodes_list)
    
    for comm_id, comm in enumerate(comms):
        for node_idx in comm:
            node_colors[node_idx] = comm_id
            
    # Calculate overall modularity Q at this step
    comm_sets = [{nodes_list[i] for i in c} for c in comms]
    Q_val = nx.community.modularity(G, comm_sets)
    
    nx.draw_networkx(
        G, pos=pos, ax=ax,
        node_color=[cmap(c) for c in node_colors],
        with_labels=True, node_size=400, font_size=9, font_weight='bold',
        edge_color='gray', alpha=0.85
    )
    ax.set_title(f"Iteration {step} ({len(comms)} Comm.)\nModularity Q = {Q_val:.4f}")
    ax.axis('off')

plt.tight_layout()
plt.show()
# Track metrics for each node across iterations (computed within their respective subgraphs)
metrics_records = []

for step, comms in enumerate(history):
    for comm_id, comm in enumerate(comms):
        sub_nodes = [nodes_list[i] for i in comm]
        sub_G = G.subgraph(sub_nodes)
        
        deg_c = nx.degree_centrality(sub_G)
        bet_c = nx.betweenness_centrality(sub_G)
        clo_c = nx.closeness_centrality(sub_G)
        clu_c = nx.clustering(sub_G)
        
        for n in sub_nodes:
            metrics_records.append({
                'Iteration': step,
                'Node': n,
                'Community': comm_id,
                'Degree': deg_c[n],
                'Betweenness': bet_c[n],
                'Closeness': clo_c[n],
                'Clustering': clu_c[n]
            })

df_metrics = pd.DataFrame(metrics_records)

# Display summary table for key hubs (e.g., Node 0 = Mr. Hi, Node 33 = Officer) and final state
print("Final Iteration Metrics (Top 10 by Degree Centrality):")
print(df_metrics[df_metrics['Iteration'] == len(history) - 1]
      .sort_values(by='Degree', ascending=False).head(10).to_string(index=False))

# Plot evolution of metrics for top influential nodes across iterations
key_nodes = [0, 1, 2, 32, 33] # Core leaders & bridges in Karate Club
metric_names = ['Degree', 'Betweenness', 'Closeness', 'Clustering']

fig, axes = plt.subplots(2, 2, figsize=(14, 9))
axes = axes.flatten()

for idx, metric in enumerate(metric_names):
    ax = axes[idx]
    for node in key_nodes:
        node_data = df_metrics[df_metrics['Node'] == node]
        ax.plot(node_data['Iteration'], node_data[metric], marker='o', linewidth=2, label=f'Node {node}')
    
    ax.set_title(f"Evolution of {metric} Centrality/Coefficient")
    ax.set_xlabel("Iteration (Split Step)")
    ax.set_ylabel(metric)
    ax.set_xticks(range(len(history)))
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend()

plt.tight_layout()
plt.show()
```
